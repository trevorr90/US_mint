```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import SplineTransformer, StandardScaler
from sklearn.linear_model import Ridge
from sklearn.pipeline import make_pipeline
from sklearn.compose import ColumnTransformer
import matplotlib.pyplot as plt
```

```python
df20 = pd.read_excel("Silver Proof Set Data Set for Interviews.xlsx", sheet_name="PY20", header=5, usecols="B:C")
df21 = pd.read_excel("Silver Proof Set Data Set for Interviews.xlsx", sheet_name="PY21", header=5, usecols="B:C")
df22 = pd.read_excel("Silver Proof Set Data Set for Interviews.xlsx", sheet_name="PY22", header=5, usecols="B:C")
df23 = pd.read_excel("Silver Proof Set Data Set for Interviews.xlsx", sheet_name="PY23", header=5, usecols="B:C")
df24 = pd.read_excel("Silver Proof Set Data Set for Interviews.xlsx", sheet_name="PY24", header=5, usecols="B:C")
df25 = pd.read_excel("Silver Proof Set Data Set for Interviews.xlsx", sheet_name="PY25", header=5, usecols="B:C")
```

```python
df20_total = df20["20RH"].sum()
df21_total = df21["21RH"].sum()
df22_total = df22["22RH"].sum()
df23_total = df23["23RH"].sum() 
df24_total = df24["24RH"].sum() 
df25_total = df25["25RH"].sum() 
print(df20_total)
print(df21_total)
print(df22_total)
print(df23_total)
print(df24_total)
print(df25_total)
```

```python
df20_total_adjusted = df20["20RH"].sum() /len(df20)
df21_total_adjusted = df21["21RH"].sum() /len(df21)
df22_total_adjusted = df22["22RH"].sum() /len(df22)
df23_total_adjusted = df23["23RH"].sum() /len(df23)
df24_total_adjusted = df24["24RH"].sum() /len(df24)
df25_total_adjusted = df25["25RH"].sum() /len(df25)
print(df20_total_adjusted)
print(df21_total_adjusted)
print(df22_total_adjusted)
print(df23_total_adjusted)
print(df24_total_adjusted)
print(df25_total_adjusted)
```

```python
#Set date column, add edition column and sales column
def prepare_coin(df, sales_col, edition):
    temp = df.copy()

    temp["Date"] = pd.to_datetime(temp["Date"])
    temp = temp.rename(columns={sales_col: "sales"})

    temp["edition"] = edition

    return temp[["Date", "Sales", "edition"]]


#Concat the datasets
coins = pd.concat([
    prepare_coin(df20, "20RH", 2020),
    prepare_coin(df21, "21RH", 2021),
    prepare_coin(df22, "22RH", 2022),
    prepare_coin(df23, "23RH", 2023),
    prepare_coin(df24, "24RH", 2024),
    prepare_coin(df25, "25RH", 2025)
], ignore_index=True)

print(coins)
```

```python
#Set launch dates so that later we can calculate days/weeks since launch
launch_dates = {
    2020: "2019-12-11",
    2021: "2020-12-09",
    2022: "2021-12-20",
    2023: "2022-11-15",
    2024: "2023-12-23",
    2025: "2024-12-01"
}

#Convert launch date to date format
launch_dates = {
    year: pd.Timestamp(date)
    for year, date in launch_dates.items()
}

#Map launch date to edition
coins["launch_date"] = coins["edition"].map(launch_dates)

print(coins)
```

```python
#Calculate days since launch for each coin
coins["days_since_launch"] = (
    coins["Date"] - coins["launch_date"]
).dt.days

#Calculate weeks since launch for each coin
coins["weeks_since_launch"] = (
    coins["days_since_launch"] // 7
)

print(coins)
```

```python
#Aggregate weekly sales
weekly = (
    coins
    .groupby(["edition", "weeks_since_launch"], as_index=False)
    ["sales"]
    .sum()
)

#Calculate each weeks percentage of that editions sales (This standardizes.)
weekly["total_edition_sales"] = (
    weekly.groupby("edition")["sales"]
    .transform("sum")
)

weekly["sales_share"] = (
    weekly["sales"] /
    weekly["total_edition_sales"]
)

print(weekly)
weekly.to_excel("Weekly.xlsx")
```

```python
#Train a model, spline and ridge regression necessary because of a lifecycle product.
X = weekly[
    ["weeks_since_launch", "edition"]
]

y = weekly["sales_share"]

preprocessor = ColumnTransformer([
    (
        "lifecycle",
        SplineTransformer(
            n_knots=8,
            degree=3
        ),
        ["weeks_since_launch"]
    ),
    (
        "edition",
        StandardScaler(),
        ["edition"]
    )
])

model = make_pipeline(
    preprocessor,
    Ridge(alpha=1.0)
)

model.fit(X, y)
```

```python
#Create the first 80 or so weeks of the 2026 coin
future_2026 = pd.DataFrame({
    "weeks_since_launch": range(0, 80),
    "edition": 2026
})

#Prediction
future_2026["predicted_sales_share"] = model.predict(
    future_2026[
        ["weeks_since_launch", "edition"]
    ]
)
#Remove negatives
future_2026["predicted_sales_share"] = (
    future_2026["predicted_sales_share"]
    .clip(lower=0)
)
```

```python
#Find predicted peak
peak = future_2026.loc[
    future_2026["predicted_sales_share"].idxmax()
]

print("Best predicted week:", peak["weeks_since_launch"])
```

```python
#Find the best 12 weeks 
top_weeks = (
    future_2026
    .nlargest(12, "predicted_sales_share")
    .sort_values("weeks_since_launch")
)

print(top_weeks)
```

```python
plt.figure(figsize=(12, 6))

plt.plot(
    future_2026["weeks_since_launch"],
    future_2026["predicted_sales_share"]
)

plt.xlabel("Weeks Since 2026 Coin Launch")
plt.ylabel("Predicted Share of Sales")
plt.title("Predicted 2026 Coin Demand Lifecycle")

plt.axvline(
    peak["weeks_since_launch"],
    linestyle="--",
    label="Predicted Peak"
)

plt.legend()
plt.show()
```
